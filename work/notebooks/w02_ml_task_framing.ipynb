## 1. My lane as an ML task (type)

*Classification, clustering, ranking, or scoring — which one, and why?*


My lane is a classification task.

The goal is to classify whether a content page is likely to experience declining search performance and should be prioritized for review.

This is a classification problem because each page is classified as either declining or not declining.

```python
print("ML Task Type: Classification")
```

## 2. Target or proxy

*What would you predict? Where does that label come from — observed outcome or a defined rule?*


The target is whether a page is declining. I create the target by labeling pages where trend_direction == "down" as 1 and all other pages as 0. This label is derived from the observed outcome in the dataset and is used as the classification target.

```python
import pandas as pd

df = pd.read_csv("data/raw/content_refresh_anonymized.csv")

df["is_declining"] = (df["trend_direction"] == "down").astype(int)

df[["trend_direction", "is_declining"]].head()
```

## 3. Success metric

*One metric you can defend. What number means 'good'?*

Precision@50 is appropriate because content teams can only review a limited number of pages. A higher Precision@50 means that more of the top 50 recommended pages actually need attention.

```python
print("Success Metric: Precision@50")
```

## 5. Why ML beats a fixed rule here

*What makes the pattern too messy for an if-statement?*

A fixed rule cannot capture the interaction between multiple SEO signals. For example, impressions, CTR, average position, content age, and search volume all influence whether a page is likely to decline. A machine learning model can learn these patterns from historical data and produce better recommendations than a single if-statement.

```python
print("Machine learning combines multiple SEO features instead of relying on a single rule.")
```
